```python
import pandas as pd

datos = {
    "clima": [
        "soleado", "soleado", "nublado", "lluvioso", "lluvioso", "lluvioso",
        "nublado", "soleado", "soleado", "lluvioso", "soleado", "nublado",
        "nublado", "lluvioso"
    ],
    "temperatura": [
        "alta", "alta", "alta", "media", "baja", "baja",
        "baja", "media", "baja", "media", "media", "media",
        "alta", "media"
    ],
    "humedad": [
        "alta", "alta", "alta", "alta", "normal", "normal",
        "normal", "alta", "normal", "normal", "normal", "alta",
        "normal", "alta"
    ],
    "viento": [
        "no", "si", "no", "no", "no", "si",
        "si", "no", "no", "no", "si", "si",
        "no", "si"
    ],
    "juega_tenis": [
        "no", "no", "si", "si", "si", "no",
        "si", "no", "si", "si", "si", "si",
        "si", "no"
    ]
}

df = pd.DataFrame(datos)

df
```

```python
X = df[["clima", "temperatura", "humedad", "viento"]]
y = df["juega_tenis"]
```

```python
# Separo las variables de entrada y la variable que queremos predecir

X = df[["clima", "temperatura", "humedad", "viento"]]
y = df["juega_tenis"]

print("Variables de entrada:")
print(X)

print("\nVariable objetivo:")
print(y)
```

```python
# Importamos las herramientas necesarias

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
# Dividimos el dataset en entrenamiento y prueba

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42
)

print("Cantidad de datos para entrenar:", len(X_train))
print("Cantidad de datos para probar:", len(X_test))
```

```python
# Convertir variables de texto en variables numéricas (preprocesamiento)

preprocesador = ColumnTransformer(
    transformers=[
        ("categoricas", OneHotEncoder(handle_unknown="ignore"), X.columns)
    ]
)
```

```python
# Creamos el modelo K-NN

modelo_knn = Pipeline(steps=[
    ("preprocesador", preprocesador),
    ("clasificador", KNeighborsClassifier(n_neighbors=3))
])

# Creamos el modelo Árbol de Decisión

modelo_arbol = Pipeline(steps=[
    ("preprocesador", preprocesador),
    ("clasificador", DecisionTreeClassifier(random_state=42))
])
```

```python
# Entrenamos los modelos

modelo_knn.fit(X_train, y_train)
modelo_arbol.fit(X_train, y_train)

print("Modelos entrenados correctamente")
```

```python
# Realizamos predicciones con los datos de prueba

pred_knn = modelo_knn.predict(X_test)
pred_arbol = modelo_arbol.predict(X_test)

print("Predicciones K-NN:")
print(pred_knn)

print("\nPredicciones Árbol de Decisión:")
print(pred_arbol)

print("\nValores reales:")
print(y_test.values)
```

```python
# Calculamos la precisión de cada modelo

precision_knn = accuracy_score(y_test, pred_knn)
precision_arbol = accuracy_score(y_test, pred_arbol)

print("Precisión K-NN:", precision_knn)
print("Precisión Árbol de Decisión:", precision_arbol)
```

```python
import matplotlib.pyplot as plt

modelos = ["K-NN", "Árbol de Decisión"]
precisiones = [precision_knn, precision_arbol]

plt.figure(figsize=(6,4))
plt.bar(modelos, precisiones)
plt.title("Comparación de precisión entre modelos")
plt.xlabel("Modelo")
plt.ylabel("Precisión")
plt.ylim(0, 1)

for i, valor in enumerate(precisiones):
    plt.text(i, valor + 0.02, str(valor), ha="center")

plt.show()
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Matriz de confusión para K-NN
ConfusionMatrixDisplay.from_predictions(y_test, pred_knn)
plt.title("Matriz de confusión - K-NN")
plt.show()
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

# Obtenemos los nombres de las variables luego del OneHotEncoder
nombres_variables = modelo_arbol.named_steps["preprocesador"].get_feature_names_out()

# Obtenemos el árbol de decisión dentro del Pipeline
arbol = modelo_arbol.named_steps["clasificador"]

plt.figure(figsize=(14,8))
plot_tree(
    arbol,
    feature_names=nombres_variables,
    class_names=arbol.classes_,
    filled=True,
    rounded=True
)

plt.title("Árbol de Decisión - Play Tennis")
plt.show()
```
